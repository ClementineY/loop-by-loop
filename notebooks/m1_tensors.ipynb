```python
import torch
print('PyTorch', torch.__version__)
print('device:', 'cuda' if torch.cuda.is_available() else 'cpu')
```

```python
x = torch.arange(6, dtype=torch.float32).reshape(2, 3)
print(x)
print('shape:', x.shape, 'dtype:', x.dtype, 'elements:', x.numel())
```

```python
# Broadcasting: center each feature across the batch
mean = x.mean(dim=0)
centered = x - mean
print('mean shape:', mean.shape)
print(centered)
```

```python
# Name axes before reducing them
images = torch.randn(8, 3, 32, 32)
channel_means = images.mean(dim=(0, 2, 3))
print('images:', images.shape, 'one mean per channel:', channel_means.shape)
```
