## Experiment question

> How does batch size change the gradient estimate?

Before running code, write a prediction. Then observe the evidence, change one variable, and explain the difference.

```python
import torch
print('PyTorch', torch.__version__)
print('device:', 'cuda' if torch.cuda.is_available() else 'cpu')
```

```python
example_gradients = torch.tensor([-6., -4., -2., -1., 0., 1., 3., 4., 5., 6., 8., 10.])
print('full-data gradient:', example_gradients.mean().item())
for batch_size in [1, 2, 4, 12]:
    print(batch_size, 'example batch estimate:', example_gradients[:batch_size].mean().item())
```

```python
from torch.utils.data import TensorDataset, DataLoader, random_split
X = torch.randn(100, 2)
y = (X[:, 0] + X[:, 1] > 0).long()
dataset = TensorDataset(X, y)
train_set, val_set, test_set = random_split(dataset, [70, 15, 15])
loader = DataLoader(train_set, batch_size=16, shuffle=True)
features, labels = next(iter(loader))
print('splits:', len(train_set), len(val_set), len(test_set))
print(features.shape, labels.shape)
```
