## Experiment question

> Can another learner reproduce your evaluation from the saved artifact?

Before running code, write a prediction. Then observe the evidence, change one variable, and explain the difference.

```python
import torch
print('PyTorch', torch.__version__)
print('device:', 'cuda' if torch.cuda.is_available() else 'cpu')
```

```python
model = torch.nn.Linear(4, 2)
fixed_input = torch.randn(3, 4)
expected = model(fixed_input).detach()
torch.save(model.state_dict(), 'model.pt')
restored = torch.nn.Linear(4, 2)
restored.load_state_dict(torch.load('model.pt', weights_only=True))
restored.eval()
actual = restored(fixed_input)
print('round trip matches:', torch.allclose(expected, actual))
```
